```python
from datetime import datetime

title = "Smart Food Management System"
print("="*40)
print(title.center(40))
print("="*40)

# inventory for food (defined before the loop so it's accessible)
inventory = [
    {
        "id": "FD001",
        "name": "Milk",
        "category": "Dairy",
        "quantity": 10,
        "unit": "litres",
        "purchase_date": "2026-10-01",
        "expiry_date": "2026-10-03"
    },
    {
        "id": "FD002",
        "name": "Rice",
        "category": "Grains",
        "quantity": 25,
        "unit": "kg",
        "purchase_date": "2026-09-15",
        "expiry_date": "2027-09-15"
    }
]

# admin and employee user name and pw
admin_username = "admin"
admin_pw = "admin123"

employee_username = "employee"
employee_pw = "employee123"

# function to add food item to inventory
def add_food_item(inventory):
  # Loop until a unique Food ID is entered
  while True:
    food_id = input("Enter Food ID (e.g. FD003) or '0' to cancel: ").strip()
    if food_id == '0':
      print("Action cancelled.")
      return
    id_exists = any(item["id"].upper() == food_id.upper() for item in inventory)
    if id_exists:
      print(f"Error: Food ID '{food_id}' already exists in the inventory. Please use a unique ID.")
    else:
      break

  name = input("Enter Food Name: ").strip().title()
  category = input("Enter Category: ").strip().title()

  # Loop until a valid integer quantity is provided
  while True:
    try:
      quantity = int(input("Enter Quantity: "))
      break
    except ValueError:
      print("Invalid input! Please enter a valid integer for Quantity.")

  unit = input("Enter Unit (e.g. kg, litres, pcs): ").strip()

  # Loop until valid purchase and expiry dates are provided
  while True:
    purchase_date_str = input("Enter Purchase Date (YYYY-MM-DD): ").strip()
    expiry_date_str = input("Enter Expiry Date (YYYY-MM-DD): ").strip()
    try:
      p_date = datetime.strptime(purchase_date_str, "%Y-%m-%d")
      e_date = datetime.strptime(expiry_date_str, "%Y-%m-%d")

      if e_date < p_date:
        print("Error: Expiry date cannot be before purchase date. Please enter valid dates.")
      else:
        purchase_date = purchase_date_str
        expiry_date = expiry_date_str
        break
    except ValueError:
      print("Error: Invalid date format! Both dates must be in YYYY-MM-DD format.")

  new_item = {
      "id": food_id.upper(),
      "name": name,
      "category": category,
      "quantity": quantity,
      "unit": unit,
      "purchase_date": purchase_date,
      "expiry_date": expiry_date
  }
  inventory.append(new_item)
  print(f"\nSuccess: {name} has been added to inventory!")

# function to search food item from inventory
def search_food_item(inventory):
  query = input("Enter Food Name or Category to search: ").strip().lower()
  found_items = []
  for item in inventory:
    if query in item["name"].lower() or query in item["category"].lower():
      found_items.append(item)

  print("\n--- Search Results ---")
  if not found_items:
    print(f"No items found matching '{query}'.")
  else:
    for item in found_items:
      print(f"ID: {item['id']} | Name: {item['name']} | Category: {item['category']} | Quantity: {item['quantity']} {item['unit']} | Purchased: {item['purchase_date']} | Expiry: {item['expiry_date']}")

# function to update food quantity in inventory
def update_quantity(inventory):
  target_id = input("Enter Food ID to update (e.g. FD001) or '0' to cancel: ").strip().upper()
  if target_id == '0':
    print("Action cancelled.")
    return

  item_found = None
  for item in inventory:
    if item["id"].upper() == target_id:
      item_found = item
      break

  if item_found:
    print(f"Current Quantity for {item_found['name']}: {item_found['quantity']} {item_found['unit']}")
    while True:
      try:
        new_qty = int(input("Enter New Quantity: "))
        if new_qty < 0:
          print("Error: Quantity cannot be negative. Please enter a valid quantity.")
        else:
          break
      except ValueError:
        print("Invalid input! Please enter a valid integer for Quantity.")
    item_found["quantity"] = new_qty
    print(f"\nSuccess: Quantity for {item_found['name']} updated to {new_qty} {item_found['unit']}.")
  else:
    print(f"\nError: Food ID '{target_id}' not found in inventory.")

# function to remove food item from inventory
def remove_food_item(inventory):
  target_id = input("Enter Food ID to remove (e.g. FD001) or '0' to cancel: ").strip().upper()
  if target_id == '0':
    print("Action cancelled.")
    return

  item_found = None
  for item in inventory:
    if item["id"].upper() == target_id:
      item_found = item
      break

  if item_found:
    inventory.remove(item_found)
    print(f"\nSuccess: {item_found['name']} has been removed from inventory!")
  else:
    print(f"\nError: Food ID '{target_id}' not found in inventory.")

# function to monitor expiry status of food items
def expiry_monitor(inventory):
  current_date = datetime.now()
  print("\n--- Expiry Status ---")
  if not inventory:
    print("No items in inventory to monitor.")
    return

  for item in inventory:
    try:
      exp_date = datetime.strptime(item["expiry_date"], "%Y-%m-%d")
      delta = (exp_date - current_date).days

      if delta < 0:
        status = "EXPIRED"
      elif 0 <= delta <= 3:
        status = "EXPIRING SOON"
      else:
        status = "AVAILABLE"

      print(f"ID: {item['id']} | Name: {item['name']} | Days to Expiry: {delta} | Status: {status}")
    except ValueError:
      print(f"ID: {item['id']} | Name: {item['name']} | Error: Invalid date format in database.")

# function to generate summary report of the inventory
def summary_report(inventory):
  current_date = datetime.now()
  total_items = len(inventory)
  total_quantity = sum(item["quantity"] for item in inventory)

  expired_count = 0
  expiring_soon_count = 0
  available_count = 0

  for item in inventory:
    try:
      exp_date = datetime.strptime(item["expiry_date"], "%Y-%m-%d")
      delta = (exp_date - current_date).days
      if delta < 0:
        expired_count += 1
      elif 0 <= delta <= 3:
        expiring_soon_count += 1
      else:
        available_count += 1
    except ValueError:
      pass

  print("\n--- Inventory Summary Report ---")
  print(f"Total Unique Items: {total_items}")
  print(f"Total Quantity of All Items: {total_quantity}")
  print(f"Expired Items: {expired_count}")
  print(f"Expiring Soon Items (0-3 days): {expiring_soon_count}")
  print(f"Available Items: {available_count}")

# asking user to give user id and pw
user_id = input("Enter your username: ").strip().lower()
user_pw = input("Enter your password: ")

# if statement for logging
if user_id == admin_username and user_pw == admin_pw:
  current_role = "ADMIN"
  print("Welcome Admin")
elif user_id == employee_username and user_pw == employee_pw:
  current_role = "EMPLOYEE"
  print("Welcome Employee")
else:
  current_role = None
  print("Invalid credentials. Access denied.")

# assigning roles and access
while current_role is not None:
  if current_role == "ADMIN":
    print("\n--- Admin menu ---")
    print("1. Add Food")
    print("2. View Inventory")
    print("3. Search Food")
    print("4. Update Quantity")
    print("5. Remove Food")
    print("6. Expiry Monitor")
    print("7. Summary")
    print("8. Logout")

    choice = input("Enter your choice: ").strip()
    if choice == "1":
      add_food_item(inventory)
    elif choice == "2":
      print("\n--- Current Inventory ---")
      if not inventory:
        print("No items in inventory.")
      else:
        for item in inventory:
          print(f"ID: {item['id']} | Name: {item['name']} | Category: {item['category']} | Quantity: {item['quantity']} {item['unit']} | Purchased: {item['purchase_date']} | Expiry: {item['expiry_date']}")
    elif choice == "3":
      search_food_item(inventory)
    elif choice == "4":
      update_quantity(inventory)
    elif choice == "5":
      remove_food_item(inventory)
    elif choice == "6":
      expiry_monitor(inventory)
    elif choice == "7":
      summary_report(inventory)
    elif choice == "8":
      print("Logging out admin...")
      current_role = None
    else:
      print(f"Selected option {choice} (Not implemented yet)")

  elif current_role == "EMPLOYEE":
    print("\n--- Employee menu ---")
    print("1. View Inventory")
    print("2. Search Food")
    print("3. Update Quantity")
    print("4. Expiry Monitor")
    print("5. Logout")

    choice = input("Enter your choice: ").strip()
    if choice == "1":
      print("\n--- Current Inventory ---")
      if not inventory:
        print("No items in inventory.")
      else:
        for item in inventory:
          print(f"ID: {item['id']} | Name: {item['name']} | Category: {item['category']} | Quantity: {item['quantity']} {item['unit']} | Purchased: {item['purchase_date']} | Expiry: {item['expiry_date']}")
    elif choice == "2":
      search_food_item(inventory)
    elif choice == "3":
      update_quantity(inventory)
    elif choice == "4":
      expiry_monitor(inventory)
    elif choice == "5":
      print("Logging out employee...")
      current_role = None
    else:
      print(f"Selected option {choice} (Not implemented yet)")
```

```python
%%writefile app.py
import streamlit as st
from datetime import datetime

st.set_page_config(page_title="Smart Food Management System", layout="centered")

# Initialize session state for inventory and login session
if "inventory" not in st.session_state:
    st.session_state.inventory = [
        {
            "id": "FD001",
            "name": "Milk",
            "category": "Dairy",
            "quantity": 10,
            "unit": "litres",
            "purchase_date": "2026-10-01",
            "expiry_date": "2026-10-03"
        },
        {
            "id": "FD002",
            "name": "Rice",
            "category": "Grains",
            "quantity": 25,
            "unit": "kg",
            "purchase_date": "2026-09-15",
            "expiry_date": "2027-09-15"
        }
    ]

if "role" not in st.session_state:
    st.session_state.role = None

# --- LOGIN SCREEN ---
if st.session_state.role is None:
    st.title("Smart Food Management System")
    st.subheader("Login")
    username = st.text_input("Username")
    password = st.text_input("Password", type="password")

    if st.button("Login"):
        if username == "admin" and password == "admin123":
            st.session_state.role = "ADMIN"
            st.rerun()
        elif username == "employee" and password == "employee123":
            st.session_state.role = "EMPLOYEE"
            st.rerun()
        else:
            st.error("Invalid credentials")

# --- MAIN DASHBOARD ---
else:
    st.title("Smart Food Management System")
    st.sidebar.write(f"Logged in as: **{st.session_state.role}**")

    if st.sidebar.button("Logout"):
        st.session_state.role = None
        st.rerun()

    menu_options = ["View Inventory", "Search Food", "Update Quantity", "Expiry Monitor"]
    if st.session_state.role == "ADMIN":
        menu_options.insert(0, "Add Food")
        menu_options.append("Remove Food")
        menu_options.append("Summary")

    choice = st.sidebar.selectbox("Navigation", menu_options)

    # 1. View Inventory
    if choice == "View Inventory":
        st.header("Current Inventory")
        if not st.session_state.inventory:
            st.info("No items in inventory.")
        else:
            st.dataframe(st.session_state.inventory)

    # 2. Add Food
    elif choice == "Add Food":
        st.header("Add New Food Item")
        with st.form("add_food_form"):
            food_id = st.text_input("Food ID (e.g. FD003)").strip().upper()
            name = st.text_input("Food Name").strip().title()
            category = st.text_input("Category").strip().title()
            quantity = st.number_input("Quantity", min_value=1, step=1)
            unit = st.text_input("Unit (e.g. kg, litres, pcs)").strip()
            p_date = st.date_input("Purchase Date")
            e_date = st.date_input("Expiry Date")
            submitted = st.form_submit_button("Add Item")

            if submitted:
                id_exists = any(item["id"] == food_id for item in st.session_state.inventory)
                if id_exists:
                    st.error(f"Error: Food ID '{food_id}' already exists.")
                elif e_date < p_date:
                    st.error("Error: Expiry date cannot be before purchase date.")
                else:
                    st.session_state.inventory.append({
                        "id": food_id,
                        "name": name,
                        "category": category,
                        "quantity": int(quantity),
                        "unit": unit,
                        "purchase_date": str(p_date),
                        "expiry_date": str(e_date)
                    })
                    st.success(f"{name} added to inventory!")

    # 3. Search Food
    elif choice == "Search Food":
        st.header("Search Food")
        query = st.text_input("Enter Food Name or Category").strip().lower()
        if query:
            results = [
                item for item in st.session_state.inventory
                if query in item["name"].lower() or query in item["category"].lower()
            ]
            if results:
                st.dataframe(results)
            else:
                st.warning(f"No items found matching '{query}'.")

    # 4. Update Quantity
    elif choice == "Update Quantity":
        st.header("Update Quantity")
        item_ids = [item["id"] for item in st.session_state.inventory]
        if not item_ids:
            st.info("Inventory is empty.")
        else:
            selected_id = st.selectbox("Select Food ID to Update", item_ids)
            target_item = next(item for item in st.session_state.inventory if item["id"] == selected_id)
            st.write(f"**Current Quantity:** {target_item['quantity']} {target_item['unit']}")
            new_qty = st.number_input("New Quantity", min_value=0, step=1, value=target_item["quantity"])
            if st.button("Update"):
                target_item["quantity"] = int(new_qty)
                st.success(f"Updated quantity for {target_item['name']} to {new_qty} {target_item['unit']}.")

    # 5. Remove Food
    elif choice == "Remove Food":
        st.header("Remove Food Item")
        item_ids = [item["id"] for item in st.session_state.inventory]
        if not item_ids:
            st.info("Inventory is empty.")
        else:
            selected_id = st.selectbox("Select Food ID to Remove", item_ids)
            if st.button("Delete Item"):
                st.session_state.inventory = [item for item in st.session_state.inventory if item["id"] != selected_id]
                st.success(f"Item {selected_id} removed.")
                st.rerun()

    # 6. Expiry Monitor
    elif choice == "Expiry Monitor":
        st.header("Expiry Status Monitor")
        current_date = datetime.now()
        monitored_list = []
        for item in st.session_state.inventory:
            exp_date = datetime.strptime(item["expiry_date"], "%Y-%m-%d")
            delta = (exp_date - current_date).days
            if delta < 0:
                status = "EXPIRED"
            elif 0 <= delta <= 3:
                status = "EXPIRING SOON"
            else:
                status = "AVAILABLE"
            monitored_list.append({
                "ID": item["id"],
                "Name": item["name"],
                "Days to Expiry": delta,
                "Status": status
            })
        st.table(monitored_list)

    # 7. Summary
    elif choice == "Summary":
        st.header("Inventory Summary Report")
        total_items = len(st.session_state.inventory)
        total_qty = sum(item["quantity"] for item in st.session_state.inventory)

        current_date = datetime.now()
        expired = 0
        expiring_soon = 0
        available = 0

        for item in st.session_state.inventory:
            delta = (datetime.strptime(item["expiry_date"], "%Y-%m-%d") - current_date).days
            if delta < 0:
                expired += 1
            elif 0 <= delta <= 3:
                expiring_soon += 1
            else:
                available += 1

        col1, col2 = st.columns(2)
        col1.metric("Total Unique Items", total_items)
        col2.metric("Total Quantity", total_qty)

        st.write("---")
        st.write(f"- **Expired Items:** {expired}")
        st.write(f"- **Expiring Soon (0–3 days):** {expiring_soon}")
        st.write(f"- **Available Items:** {available}")

```
